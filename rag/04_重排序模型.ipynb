### 重排序

```python
query = "什么是文本排序模型"
documents = [
            "文本排序模型广泛用于搜索引擎和推荐系统中，它们根据文本相关性对候选文本进行排序",
            "量子计算是计算科学的一个前沿领域",
            "预训练语言模型的发展给文本排序模型带来了新的进展"
        ]
```

```python
def get_qwen3_rerank_results(query:str,documents:list) -> list:
    response = dashscope.TextReRank.call(
        model="qwen3-rerank",
        query=query,
        documents=documents,
        top_n=3,
        return_documents=True,
        api_key=os.environ["DASHSCOPE_API_KEY"]
    )
    
    if response.status_code ==HTTPStatus.OK:
        return response["output"]["results"]  #每个结果都包含index，相关性分数，文档内容
    else:
        return response.text
```

```python
get_qwen3_rerank_results(query=query,documents=documents)
```

```python
def get_gte_rerank_v2(query:str,documents:list) -> list:
    response = dashscope.TextReRank.call(
        model="gte-rerank-v2",
        query=query,
        documents=documents,
        top_n=3,
        return_documents=True,
        api_key=os.environ["DASHSCOPE_API_KEY"]
    )
    
    if response.status_code ==HTTPStatus.OK:
        return response["output"]["results"]  #每个结果都包含index，相关性分数，文档内容
    else:
        return response.message
```

```python
get_gte_rerank_v2(query=query,documents=documents)
```
