```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
#chunk：创建切分器
spliter = RecursiveCharacterTextSplitter(chunk_size=128,chunk_overlap=64)
```

```python
text = """

0. **定义**

大模型一般指**大语言模型（Large Language Model, LLM）**，是基于Transformer架构、通过海量文本数据训练的深度学习模型，参数量通常达到亿级甚至万亿级，具备强大的自然语言理解与生成能力。广义上也包含多模态大模型（融合文本、图像、音频等信息）。

1. **核心特征**

    - 参数量巨大：从十亿级（如Llama 7B）到万亿级（如GPT-4）不等。

    - 海量训练数据：涵盖书籍、网页、论文等各类文本语料。

    - 涌现能力：模型规模达到阈值后，会出现小模型不具备的能力（如逻辑推理、代码生成）。

    - 上下文理解：能够基于输入的上下文进行连贯的文本生成与对话。

2. **关键术语**

    - 预训练：模型在海量无标注数据上学习语言规律的过程，是大模型能力的基础。

    - 微调（Fine-tuning）：在预训练模型基础上，使用特定任务的标注数据进行训练，适配下游任务。

    - 提示工程（Prompt Engineering）：通过设计合理的输入提示词，引导模型生成预期输出。

    - 上下文窗口：模型能够处理的输入文本最大长度（如GPT-3.5为4k，GPT-4 Turbo为128k）。

    - 幻觉：模型生成看似合理但与事实不符的内容，是大模型的核心缺陷之一。

"""
```

```python
texts = spliter.split_text(text=text)
```
