## 向量化模型

```python
def get_qwen3_embeddings(texts:list):
    response=dashscope.TextEmbedding.call(
        model="text-embedding-v4",
        input=texts,
        api_key=os.environ["DASHSCOPE_API_KEY"]
    )
    results = [item["embedding"] for item in response.output["embeddings"]]
    
    return results
```

```python
texts = ["你好","睡觉"]
```

```python
embeddings = get_qwen3_embeddings(texts)
```
