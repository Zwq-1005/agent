```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_messages(
    messages=[
        ("system","你是一个语言翻译专家，请把用户输入的内容翻译成{language}"),
        ("user","{text}")
    ]
)
```

```python
prompt
```

```python
language="法语"
text = "檀健次是世界上最帅的男人！"
```

```python
message = prompt.invoke(input=dict(language=language,text=text))
```

```python
from models import get_qwen_llm
```

```python
llm = get_qwen_llm()
```

```python
response = llm.invoke(message)
```

```python
response
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
str_parser = StrOutputParser()
```

```python
str_parser.invoke(response)
```

```python
#构建chain
translate_chain = prompt|llm|str_parser
```

```python
translate_chain.invoke(input=dict(language=language,text=text))
```
