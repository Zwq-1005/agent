### 1.引入模型

```python
from models import get_qwen_llm
```

```python
llm=get_qwen_llm()
```

### 2.构建消息

```python
from langchain.messages import SystemMessage
from langchain.messages import HumanMessage
from langchain.messages import AIMessage
```

```python
message1 = [
    SystemMessage(content="你是一个语言翻译专家，请把用户输入的内容翻译成英语"),
    HumanMessage(content="檀健次是世界上最帅的男人！")
]

message2 = [
    HumanMessage(content="感冒了怎么办？")
]
```

### 3.调用

```python
#单样本阻塞式
llm.invoke(input=message1)
```

```python
llm.stream(input=message1)
```

```python
for chunk in llm.stream(input=message2):
    print(chunk.content,sep="",end="")
```

```python
#批量调用
llm.batch(inputs=[message1,message2])
```
